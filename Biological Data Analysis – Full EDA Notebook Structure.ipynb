# Dataset Short Summary

This dataset contains 10,000+ biological records.

This is a structured dataset created for biomedical / healthcare predictive analysis.

# Features:

    Sample_ID                 – An ID that uniquely identifies each biological sample

    Species_Name              – The species to which the sample belongs

    Gene_Expression_Value     – A measure of gene activity intensity

    Protein_Level             – A measure of protein concentration

    Disease_Status            – Healthy / Diseased (Target Variable)

    Experimental_Conditions   – Test conditions such as control / stress / treatment

# Dataset Nature

Mixed Data Types (Numerical + Categorical)

Classification Problem (Disease Prediction)

Gene & Protein biological indicators

Experimental condition impact analysis dataset

# Step 1:Import Libraries

```python
# Data Handling
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Settings
plt.figure(figsize=(8,6))
sns.set(style="whitegrid")
```

# Step 2:Load Dataset

```python
df = pd.read_csv("biological_dataset_10000_records.csv")

df.head()
```

# Step 3: Dataset Overview

```python
print("Shape:", df.shape)
print("\nData Types:\n")
print(df.dtypes)
```

```python
print("\nDataset Info:\n")
df.info()
```

```python
print("\nStatistical Summary:\n")
df.describe()
```

✔ Total records

✔ Numerical summary

✔ Data type validation

# Step 4:Missing Value Analysis

```python
df.isnull().sum()
```

Visualize Missing Values

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing Values Heatmap")
plt.show()
```

# Step 5:Duplicate Check

```python
print("Duplicate Records:", df.duplicated().sum())
```

# Step 6:Class Distribution Analysis

```python
df['Disease_Status'].value_counts()
```

Visualization

```python
plt.figure()
sns.countplot(x='Disease_Status', data=df)
plt.title("Disease Status Distribution")
plt.show()
```

✔ Check class imbalance

# Step 7:Univariate Analysis – Numerical Features

Gene Expression Distribution

```python
plt.figure()
plt.hist(df['Gene_Expression_Value'], bins=30)
plt.title("Gene Expression Distribution")
plt.xlabel("Gene Expression Value")
plt.ylabel("Frequency")
plt.show()
```

Protein Level Distribution

```python
plt.figure()
plt.hist(df['Protein_Level'], bins=30)
plt.title("Gene Expression Distribution")
plt.xlabel("Protein_Level")
plt.ylabel("Frequency")
plt.show()
```

✔ Skewness

✔ Spread

✔ Outliers

# Step 8:Boxplot – Outlier Detection

```python
plt.figure()
sns.boxplot(x=df['Gene_Expression_Value'])
plt.title("Gene Expression Outliers")
plt.show()

plt.figure()
sns.boxplot(x=df['Protein_Level'])
plt.title("Protein Level Outliers")
plt.show()
```

# Step 9:Disease vs Feature Relationship

```python
plt.figure()
sns.boxplot(x='Disease_Status', 
            y='Gene_Expression_Value', 
            data=df)
plt.title("Gene Expression vs Disease")
plt.show()
```

Protein Level vs Disease

```python
plt.figure()
sns.boxplot(x='Disease_Status', 
            y='Protein_Level', 
            data=df)
plt.title("Protein Level vs Disease")
plt.show()
```

✔ Feature separation visible?

# Step 10:Correlation Analysis

```python
numeric_df = df.select_dtypes(include=[np.number])

# Correlation
corr = numeric_df.corr()

plt.figure(figsize=(6,5))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

If you know the numeric columns in your dataset:

```python
corr = df[['Gene_Expression_Value','Protein_Level']].corr()

plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

✔ Multicollinearity

✔ Strong relationships

# Step 1.1:Species-wise Analysis

```python
plt.figure(figsize=(8,6))
sns.countplot(x='Species_Name', 
              hue='Disease_Status', 
              data=df)
plt.xticks(rotation=45)
plt.title("Species vs Disease Status")
plt.show()
```

✔ Species susceptibility

| Skewness Value | Meaning                |
